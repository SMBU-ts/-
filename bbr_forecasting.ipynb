# BBR Forecasting: VAR and VECM

This notebook runs the experiment for `D:\\ALL\\bbr` using two target variables: `rtt` and `cwnd`.

It compares two models under two window strategies:

- VAR
- VECM
- Expanding window
- Sliding window

Outputs are written to `D:\\ALL\\results_bbr` by default.

## 2. Import Experiment Code

The implementation lives in `run_bbr_forecasting.py`, and this notebook controls the workflow.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import Image, display

ROOT = Path(r"D:\ALL")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from run_bbr_forecasting import (
    ExperimentConfig,
    configure_logging,
    run_forecasts,
    write_outputs,
)
```

## 3. Quick Smoke Test

This small run checks that the notebook, dependencies, data loading, models, metrics, and plots work. The values from this test are not the final thesis results.

```python
smoke_config = ExperimentConfig(
    data_dir=ROOT / "bbr",
    output_dir=ROOT / "results_bbr_smoke_notebook",
    max_files=1,
    test_steps=2,
)

configure_logging(smoke_config.output_dir)
smoke_predictions, smoke_metrics = run_forecasts(smoke_config)
write_outputs(smoke_predictions, smoke_metrics, smoke_config)
smoke_metrics
```

## 4. Full Experiment

This cell runs all 50 BBR CSV files, with 720 test steps and both window methods. It may take a long time because VAR and VECM are refit for every test step.

```python
full_config = ExperimentConfig(
    data_dir=ROOT / "bbr",
    output_dir=ROOT / "results_bbr",
    train_size=2880,
    test_size=720,
    sliding_size=720,
    maxlags=10,
    seed=42,
)

configure_logging(full_config.output_dir)
predictions, metrics = run_forecasts(full_config)
write_outputs(predictions, metrics, full_config)
metrics
```
